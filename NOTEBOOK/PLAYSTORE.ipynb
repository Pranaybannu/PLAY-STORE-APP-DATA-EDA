```python
##adjusting the row and column values

app.loc[
    app['App'] == 'Life Made WI-Fi Touchscreen Photo Frame',
    'Category']=np.nan

app.loc[
    app['App'] == 'Life Made WI-Fi Touchscreen Photo Frame', 'Rating']=1.9

app.loc[
    app['App'] == 'Life Made WI-Fi Touchscreen Photo Frame','Reviews'] = '19.0'

app.loc[
    app['App'] == 'Life Made WI-Fi Touchscreen Photo Frame','Size']='3.0M'

app.loc[
    app['App'] == 'Life Made WI-Fi Touchscreen Photo Frame','Installs']='1,000+'

app.loc[
    app['App'] == 'Life Made WI-Fi Touchscreen Photo Frame','Type']='Free'

app.loc[
    app['App'] == 'Life Made WI-Fi Touchscreen Photo Frame','Price']='0.0'

app.loc[
    app['App'] == 'Life Made WI-Fi Touchscreen Photo Frame','Content Rating']='Everyone'

app.loc[
    app['App'] == 'Life Made WI-Fi Touchscreen Photo Frame','Genres']=np.nan

app.loc[
    app['App'] == 'Life Made WI-Fi Touchscreen Photo Frame','Last Updated']='February 11, 2018'

app.loc[
    app['App'] == 'Life Made WI-Fi Touchscreen Photo Frame','Current Ver']='1.0.19'

app.loc[
    app['App'] == 'Life Made WI-Fi Touchscreen Photo Frame','Android Ver']='4.0 and up'
```

```python
app[app['Last Updated']=='February 11, 2018']
```

```python
app['updated'] = pd.to_datetime(app['Last Updated'],format='mixed')
```

```python
app.head()
```

```python
## extracting year and month from updated column
app['year']= app['updated'].dt.year
```

```python
app['month']=app['updated'].dt.month_name()
```

```python
app.head()
```

```python
app['Genres'].unique()
```

```python
def prim_gen(x):
    if not isinstance(x, str):
        return x                                       ##for primary genres
    a = x.find(';')
    if a != -1:
        return x[:a]
    return x
```

```python
##creating primary genre column
app['primary_genre']= app['Genres'].apply(prim_gen)
```

```python
app.head()
```

```python
app.info()
```

```python
##converting reviews to number/int/float
app['rev_num'] = pd.to_numeric(app['Reviews'],errors='coerce')
```

```python
app.head()
```

```python
app['Installs'].min(),app['Installs'].max()
```

```python
app[app['Installs']=='0']  ##only one app with zero installs
```

```python
##converting app size to numeric
def siz(x):
  if 'M' in x:
    return float(x.replace('M',''))
  elif 'k' in x:
    return float(x.replace('k',''))
  else:
    return np.nan
```

```python
app['size_mb']=app['Size'].apply(siz)
```

```python
app.head()
```

```python
app['size_mb'].value_counts().head()
```

```python
app['Price'].unique()
```

```python
def pric(x):
  if '$' in x:
    return float(x.replace('$',''))
  else:
    return float(x)
```

```python
##converting price to numeric
app['price_dol'] = app['Price'].apply(pric)
```

```python
app['price_dol'].unique()
```

#### Chart - 12 Multivariable analysis

```python
rev.info()
```

```python
## polarit bucket
data = rev[['Sentiment_Polarity','Sentiment_Subjectivity','review_len']]
```

```python
## polarit bucket and subjectivity buckets
def pol(x):
  if x<0:
    if x>-0.3 and x<0:
      return 'Low-negative-polarity'
    elif x>-0.7 and x<-0.3:
      return 'Moderate-negative-polarity'
    else:
      return 'High-negative-polarity'
  else:
    if x>0 and x<0.3:
      return 'Low-Positive-polarity'
    elif x>0.3 and x<0.7:
      return 'Moderate-positive-polarity'
    else:
      return 'High-positive-polarity'



def sub(x):
  if x==0:
    return 'objective'
  elif x>0 and x<0.3:
    return 'Low-subjectivity'
  elif x>0.3 and x<0.7:
    return 'Moderate-subjectivity'
  else:
    return 'High-subjectivity'
```

```python
data['pol_buc'] = data['Sentiment_Polarity'].apply(pol)
data['sub_buc'] = data['Sentiment_Subjectivity'].apply(sub)
```

```python
data.head()
```

```python
# Chart - 12 visualization code
fig = make_subplots(
    rows=2, cols=2,
    subplot_titles=(
        'Box plot for sentiment polarity distribution wrt sentiment',
        'Box plot for sentiment subjectivtiy distribution wrt sentiment',
        'Barplot for sentiment polarity count',
        'Barplot for sentiment subjectivity count'
    )
)

# box
fig.add_trace(
    go.Box(x=rev['Sentiment'],y=rev['Sentiment_Polarity'],name='polarity vs sentiment'),
    row=1, col=1)

# box
fig.add_trace(
    go.Box(x=rev['Sentiment'],y=rev['Sentiment_Subjectivity'], name='subjectivity vs sentiment'),
    row=1, col=2)

# Barplot
df = data['pol_buc'].value_counts().reset_index()
df.sort_values(by='count',ascending=True,ignore_index=True,inplace=True)
fig.add_trace(
     go.Bar(x=df['pol_buc'],y=df['count'],name='polarity count'),
    row=2,col=1)

df1 = data['sub_buc'].value_counts().reset_index()
df1.sort_values(by='count',ascending=True,ignore_index=True,inplace=True)
fig.add_trace(
     go.Bar(x=df1['sub_buc'],y=df1['count'],name='subjectivity count'),
    row=2,col=2)

fig.update_layout(
    title='Subplots',
    title_font_size=20,
    width=1400,
    title_font_weight = 'bold',
    font= dict(size=12,weight='bold')
)
```
